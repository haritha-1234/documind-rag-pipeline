
# NOTEBOOK 01 — DOCUMENT INGESTION

 Ingestion is the very first step of the RAG pipeline.cIt means: loading your raw document (PDF, txt, etc.) into Python so we can work with it. Think of it like opening a book before you can read it. Without this step, the pipeline has nothing to work with.

In this notebook we will:
- Load a PDF document using LangChain's PDF loader
- See what the loaded document looks like
- Understand the structure LangChain uses

```python
# Import the PDF loader from LangChain
# PyPDFLoader knows how to open and read PDF files
# It converts each page into a LangChain "Document" object
from langchain_community.document_loaders import PyPDFLoader

# os helps us work with file paths
import os

print("Libraries imported successfully!")
```

```python
# ============================================================
# STEP 1: LOAD THE PDF
# ============================================================

# Define the path to our PDF file
pdf_path = "../data/documents/attention_is_all_you_need.pdf"

# Create a PyPDFLoader object — this is our document reader
# Think of it like telling Python: "here is the book I want to open"
loader = PyPDFLoader(pdf_path)

# .load() actually reads the PDF and returns a list of Documents
# Each page in the PDF becomes one Document object
documents = loader.load()

print(f"Document loaded successfully!")
print(f"Total pages loaded: {len(documents)}")
```

```python
# ============================================================
# STEP 2: UNDERSTAND THE STRUCTURE
# ============================================================

# LangChain wraps each page in a "Document" object
# Every Document has two parts:
#   - page_content : the actual text on that page
#   - metadata     : extra info like page number, source file

# Let's look at the first page
first_page = documents[0]

print("=== TYPE OF OBJECT ===")
print(type(first_page))  # Should be langchain Document

print("\n=== METADATA ===")
print(first_page.metadata)  # Shows page number and source path

print("\n=== FIRST 500 CHARACTERS OF TEXT ===")
print(first_page.page_content[:500])  # Just a preview
```

```python
# ============================================================
# STEP 3: EXPLORE A FEW PAGES
# ============================================================

# Let's loop through the first 3 pages to see the structure
for i, doc in enumerate(documents[:3]):

    print(f"\n{'='*50}")
    print(f"PAGE {i+1}")
    print(f"{'='*50}")

    # Show the metadata for this page
    print(f"Metadata : {doc.metadata}")

    # Show first 300 characters of content
    # We slice with [:300] to avoid printing the whole page
    print(f"Preview  : {doc.page_content[:300]}")
```

```python
# ============================================================
# SUMMARY — WHAT DID WE LEARN?
# ============================================================

print("=== INGESTION SUMMARY ===")
print(f"Document loaded from  : {pdf_path}")
print(f"Total pages           : {len(documents)}")
print(f"Type of each page     : {type(documents[0])}")
print(f"Keys in metadata      : {list(documents[0].metadata.keys())}")

print("""
WHAT HAPPENS NEXT?
------------------
Now that we have loaded the document into memory as a list
of Document objects, the next step is CHUNKING.

Chunking means splitting each page into smaller pieces
so the LLM can focus on just the relevant part
when answering a question.

See notebook 02_chunking.ipynb for the next step!
""")
```
